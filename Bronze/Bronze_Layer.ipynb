### Reading bulk CSV files and writing them into Bronze tables

```python
INGESTION_CONFIG = [
    {
        #"source": "crm",
        "path": "/Volumes/data_lakehouse/bronze/file_storage/source_crm/prd_info.csv",
        "table": "crm_prd_info"
    },
    {
        #"source": "crm",
        "path": "/Volumes/data_lakehouse/bronze/file_storage/source_crm/sales_details.csv",
        "table": "crm_sales_details"
    },
    {
        #"source": "erp",
        "path": "/Volumes/data_lakehouse/bronze/file_storage/source_erp/CUST_AZ12.csv",
        "table": "erp_cust_az12"
    },
    {
        #"source": "erp",
        "path": "/Volumes/data_lakehouse/bronze/file_storage/source_erp/LOC_A101.csv",
        "table": "erp_loc_a101"
    },
    {
        "source": "erp",
        "path": "/Volumes/data_lakehouse/bronze/file_storage/source_erp/PX_CAT_G1V2.csv",
        "table": "erp_px_cat_g1v2"
    }
]
```

```python
for item in INGESTION_CONFIG:
    print(f"Ingesting : data_lakehouse.bronze.{item['table']}")

    df = (
        spark.read
             .option("header", "true")
             .option("inferSchema", "true")
             .csv(item["path"])
    )

    (
        df.write
          .mode("overwrite")
          .format("delta")
          .saveAsTable(f"data_lakehouse.bronze.{item['table']}")
    )
```
