### Table Name : gold_dim_customers

```python
query = """
SELECT
    ROW_NUMBER() OVER (ORDER BY ci.customer_id) AS customer_key,
    ci.customer_id,
    ci.customer_number,
    ci.first_name,
    ci.last_name,
    la.CNTRY,
    ci.marital_status,
    CASE
        WHEN ci.gender <> 'N/A' THEN ci.gender
        ELSE COALESCE(ca.GEN, 'N/A')
    END AS gender,
    ca.bdate AS birthdate,
    ci.created_date AS create_date
FROM data_lakehouse.silver.crm_cust_info ci
LEFT JOIN data_lakehouse.silver.erp_cust_az12 ca
    ON ci.customer_number = ca.CID
LEFT JOIN data_lakehouse.silver.erp_loc_a101 la
    ON ci.customer_number = la.CID
"""

df = spark.sql(query)
```

### Writing to GOLD table

```python
df.write.mode('overwrite').format('delta').saveAsTable('data_lakehouse.gold.gold_dim_customers')
```
