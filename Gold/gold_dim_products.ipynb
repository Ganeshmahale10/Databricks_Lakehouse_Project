### Table Name : gold.dim_products

```python
query = """
SELECT
    ROW_NUMBER() OVER (ORDER BY pn.start_date, pn.product_number) AS product_key, -- Surrogate key
    pn.product_id,
    pn.product_number,
    pn.product_name,
    pn.category_id,
    pc.cat,
    pc.subcat,
    pc.maintenance,
    pn.product_cost,
    pn.product_line,
    pn.start_date
FROM data_lakehouse.silver.crm_prd_info pn
LEFT JOIN data_lakehouse.silver.erp_px_cat_g1v2 pc
    ON pn.category_id = pc.id
"""

df = spark.sql(query)
```

```python
df.write.mode("overwrite").format('delta').saveAsTable("data_lakehouse.gold.gold_dim_products")
```
