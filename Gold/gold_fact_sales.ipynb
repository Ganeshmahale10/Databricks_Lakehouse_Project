### Table Name : gold.fact_sales

```python
query = """
SELECT
    sd.sales_order_number  AS order_number,
    pr.product_key  AS product_key,
    cu.customer_key AS customer_key,
    sd.sales_order_date AS order_date,
    sd.sales_ship_date  AS shipping_date,
    sd.sales_due_date   AS due_date,
    sd.sales_amount  AS sales_amount,
    sd.sales_quantity AS quantity,
    sd.sales_price    AS price
FROM data_lakehouse.silver.crm_sales_info sd
LEFT JOIN data_lakehouse.gold.gold_dim_products pr
    ON sd.sales_product_key = pr.product_number
LEFT JOIN data_lakehouse.gold.gold_dim_customers cu
    ON sd.sales_customer_id = cu.customer_id;
"""

df = spark.sql(query)
```

```python
df.write.mode("overwrite").format("delta").saveAsTable("data_lakehouse.gold.gold_fact_sales")
```
