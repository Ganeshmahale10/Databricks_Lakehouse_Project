### Table Name : erp_cust_az12

```python
from pyspark.sql.functions import *
```

```python
df = spark.read.table('data_lakehouse.bronze.erp_cust_az12')
```

###Triming Spaces

```python
for fields in df.schema.fields:
    if isinstance(fields.dataType,StringType):
        df = df.withColumn(fields.name,trim(fields.name))
```

### Extracting customer ID

```python
df = df.withColumn('CID',substring('CID',4,length('CID')))

```

```python
df = df.withColumn('GEN',
                   when(upper(col('GEN')).isin('M','MALE'),'Male')
                   .when(upper(col('GEN')).isin('F', 'FEMALE'),'Female')
                   .otherwise('NA'))

```

```python
df.write.mode('overwrite').format('Delta').saveAsTable('data_lakehouse.silver.erp_cust_az12')
```
