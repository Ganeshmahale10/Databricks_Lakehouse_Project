### Table Name : erp_loc_a101

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df = spark.read.table('data_lakehouse.bronze.erp_loc_a101')
```

```python
df = df.withColumn(
    'CNTRY',
    when(upper(col('CNTRY')).isin('US', 'USA'), 'United States')
    .when(upper(col('CNTRY')) == 'DE', 'Germany')
    .when(col('CNTRY').isNull() | (col('CNTRY') == ' '), 'NA')
    .otherwise(col('CNTRY'))
)
```

```python
for fields in df.schema.fields:
    if isinstance(fields.dataType,StringType):
        df.withColumn('fields.name',trim(fields.name))
```

```python
df = df.withColumn('CID',regexp_replace('CID','-',''))
```

```python
df.write.mode('overwrite').format('delta').saveAsTable('data_lakehouse.silver.erp_loc_a101')
```
